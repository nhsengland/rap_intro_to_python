# 🕒 Python `datetime` Mini Workshop

Learn how to work with dates, times, and time differences in Python.

```python
# ✅ 1. Importing the datetime module

import datetime

```

```python
# 2️⃣ Create a date object for today
today = datetime.date.today()
print("Today's date:", today)
```

```python
# ✅ Check the type of the date object
type(today)
```

```python
# 3️⃣ Create a specific date
birthday = datetime.date(1995, 5, 17)
print("Birthday:", birthday)
```

```python
# ✅ Type of the specific date
type(birthday)
```

```python
# 4️⃣ Access components of a date
print("Year:", birthday.year)
print("Month:", birthday.month)
print("Day:", birthday.day)
```

```python
# ✅ Type of the year component
type(birthday.year)
```

```python
# 5️⃣ Create a time object
t = datetime.time(14, 30, 45)
print("Time:", t)
```

```python
# ✅ Type of the time object
type(t)
```

```python
# 6️⃣ Access components of time
print("Hour:", t.hour)
print("Minute:", t.minute)
print("Second:", t.second)
```

```python
# Type of the hour component
type(t.hour)
```

```python
# 7️⃣ Create a datetime object (date + time)
now = datetime.datetime.now()
print("Current datetime:", now)
```

```python
# ✅ Type of datetime object
type(now)
```

```python
# 8️⃣ Create a specific datetime
meeting = datetime.datetime(2025, 10, 21, 9, 30)
print("Meeting datetime:", meeting)
```

```python
# 9️⃣ Access date and time parts separately
print("Date part:", meeting.date())
print("Time part:", meeting.time())
```

```python
# 1️⃣0️⃣ Formatting datetime as string (strftime)
formatted = now.strftime("%A, %B %d, %Y at %I:%M %p")
print("Formatted datetime:", formatted)
```

```python
# 1️⃣1️⃣ Parsing a string to datetime (strptime)
date_string = "2025-10-21 09:30"
parsed = datetime.datetime.strptime(date_string, "%Y-%m-%d %H:%M")
print("Parsed datetime:", parsed)
```

```python
# ✅ Type of parsed datetime
type(parsed)
```

```python
# 1️⃣2️⃣ Timedelta (date arithmetic)
delta = datetime.timedelta(days=10)
print("Timedelta:", delta)
```

```python
# ✅ Type of timedelta
type(delta)
```

```python
# Add or subtract timedelta
future_date = today + delta
past_date = today - delta

print("10 days from today:", future_date)
print("10 days before today:", past_date)
```

```python
# 1️⃣3️⃣ Comparing dates
deadline = datetime.date(2025, 12, 31)
if today < deadline:
    print("Still time before the deadline!")
else:
    print("Deadline passed.")
```
