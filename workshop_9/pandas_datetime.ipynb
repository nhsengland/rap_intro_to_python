## 1. Importing Pandas

We use Pandas for powerful datetime handling in series and DataFrames.  
Let's import pandas as `pd`.


```python
import pandas as pd
```

## 2. Create a Series of Dates

Create a series of date strings and see how Pandas handles them.


```python
dates = pd.Series(['2025-01-01', '2025-02-14', '2025-12-31'])
print(dates)
```

```python
# Check the type of the series and its first element
print(type(dates))
print(type(dates[0]))
```

## 3. Convert Strings to Datetime

Pandas allows converting strings to datetime objects using `pd.to_datetime`.


```python
dates_dt = pd.to_datetime(dates)
print(dates_dt)
```

```python
# Check type of first element
print(type(dates_dt[0]))
```

## 4. Accessing Date Components

Once we have datetime objects, we can extract **year**, **month**, **day**, and more.


```python
print("Year:", dates_dt.dt.year)
print("Month:", dates_dt.dt.month)
print("Day:", dates_dt.dt.day)
```

## 5. Date Arithmetic

We can add or subtract time using `pd.Timedelta`.


```python
delta = pd.Timedelta(days=10)
future_dates = dates_dt + delta
past_dates = dates_dt - delta

print("Future dates:\n", future_dates)
print("Past dates:\n", past_dates)
```

```python
# Check type of delta
print(type(delta))
```

## 6. Filtering Dates

We can filter datetime series using comparison operators.


```python
# Filter dates after 2025-02-01
filtered = dates_dt[dates_dt > '2025-02-01']
print(filtered)
```

## 7. Day of Week and Month Name

Pandas can extract **day of week** and **month name** easily.


```python
print("Day of week:", dates_dt.dt.day_name())
print("Month name:", dates_dt.dt.month_name())
```
